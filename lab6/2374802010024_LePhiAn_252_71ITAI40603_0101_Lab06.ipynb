# LAB 06: NHẬN DIỆN ẢNH (IMAGE DETECTION)


```python
import cv2
import matplotlib.pyplot as An

img = cv2.imread('barbara.png')
img2 = cv2.imread('zelda.png')

def show_img(image, title='Image'):
    if image is None: return
    if len(image.shape) == 3:
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    else:
        image_rgb = image
    An.imshow(image_rgb, cmap='gray' if len(image.shape) == 2 else None)
    An.title(title)
    An.axis('off')
    An.show()
```

## 1. Đoạn code mẫu: Nhận diện với Stop Sign/Face


```python
if img is not None:
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img_rgb = img.copy()
    stop_data = cv2.CascadeClassifier('stop_data.xml')
    if not stop_data.empty():
        found = stop_data.detectMultiScale(img_gray, minSize=(20, 20))
        if len(found) != 0:
            for (x, y, width, height) in found:
                cv2.rectangle(img_rgb, (x, y), (x + height, y + width), (0, 255, 0), 5)
    show_img(img_rgb, 'Sample Code Detection')
```

## 2. Code nhận diện Khuôn mặt (Face Detection)


```python
if img is not None:
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img_face = img.copy()
    face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')
    if not face_cascade.empty():
        faces = face_cascade.detectMultiScale(img_gray, scaleFactor=1.05, minNeighbors=3, minSize=(30, 30))
        for (x, y, w, h) in faces:
            cv2.rectangle(img_face, (x, y), (x + w, y + h), (255, 0, 0), 2)
    show_img(img_face, 'Face Detection')
```

## 3. Code nhận diện Mắt (Eye Detection)


```python
if img2 is not None:
    img_gray = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
    img_eye = img2.copy()
    eye_cascade = cv2.CascadeClassifier('haarcascade_eye_tree_eyeglasses.xml')
    if not eye_cascade.empty():
        eyes = eye_cascade.detectMultiScale(img_gray, scaleFactor=1.05, minNeighbors=3, minSize=(10, 10))
        for (x, y, w, h) in eyes:
            cv2.rectangle(img_eye, (x, y), (x + w, y + h), (0, 0, 255), 2)
    show_img(img_eye, 'Eye Detection')
```

## 4. Code nhận diện Nụ cười (Smile Detection)


```python
if img2 is not None:
    img_gray = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
    img_smile = img2.copy()
    smile_cascade = cv2.CascadeClassifier('haarcascade_smile.xml')
    if not smile_cascade.empty():
        smiles = smile_cascade.detectMultiScale(img_gray, scaleFactor=1.1, minNeighbors=10)
        for (x, y, w, h) in smiles:
            cv2.rectangle(img_smile, (x, y), (x + w, y + h), (0, 255, 255), 2)
    show_img(img_smile, 'Smile Detection')
```
