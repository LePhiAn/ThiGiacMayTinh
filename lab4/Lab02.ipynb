```python
import cv2
import matplotlib.pyplot as PAn
```

```python
img = cv2.imread('../img/cat.jpg')
img = img[:,:,::-1]  # BGR to RGB
PAn.imshow(img)
```

```python
blur_3 = cv2.blur(img, (3,3))
blur_7 = cv2.blur(img, (7,7))
```

```python
PAn.figure(figsize=(10,4))

PAn.subplot(1,3,1)
PAn.imshow(img)
PAn.title("Original")
PAn.axis("off")

PAn.subplot(1,3,2)
PAn.imshow(blur_3)
PAn.title("Box 3x3")
PAn.axis("off")

PAn.subplot(1,3,3)
PAn.imshow(blur_7)
PAn.title("Box 7x7")
PAn.axis("off")

PAn.show()
```

## Nhận xét

- Kích thước kernel ảnh hưởng trực tiếp đến mức độ làm mờ của ảnh.
- Kernel nhỏ (3×3): giảm nhiễu ít, giữ được nhiều chi tiết.
- Kernel trung bình (5×5): cân bằng giữa khử nhiễu và bảo toàn biên.
- Kernel lớn (7×7): khử nhiễu mạnh nhưng làm mờ các chi tiết nhỏ.
- Bộ lọc Box và Gaussian không phù hợp để xử lý nhiễu pepper do cơ chế lấy trung bình các pixel.
- Bộ lọc Median cho kết quả tốt nhất khi xử lý nhiễu pepper, đặc biệt với kernel 5×5 hoặc 7×7.


```python
img1 = cv2.imread('../img/exT3_01.png')
img2 = cv2.imread('../img/exT3_02.png')
img1 = img1[:155,1:191]
img2 = img2[:155,:190]
inv_img1 = cv2.bitwise_not(img1)
inv_img2 = cv2.bitwise_not(img2)
```

```python
PAn.subplot(221)
PAn.imshow(img1[:,:,::-1])
PAn.subplot(222)
PAn.imshow(img2[:,:,::-1])
PAn.subplot(223)
PAn.imshow(inv_img1[:,:,::-1])
PAn.subplot(224)
PAn.imshow(inv_img2[:,:,::-1])
```

```python
# Tạo ball

gray = cv2.cvtColor(inv_img1, cv2.COLOR_BGR2GRAY)

_, binary = cv2.threshold(
    gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5,5))
ball = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
```

```python
PAn.imshow(ball, cmap='gray')
```
