#  Geometric Transformations


 Geometric transformations allow you to perform different operations like translation i.e. to shift, reshape and rotate the image.


## Scaling 


We can resize an image using the function `resize()` from `cv2` module for this purpose.  You can specify the scaling factor or the size of the image:


Consider the following image with the corresponding intensity values:


```python
toy_image = np.zeros((6,6)) # tạo hình ảnh đen 6x6
toy_image[1:5,1:5]=255 # tạo hình ảnh trắng bên trong
toy_image[2:4,2:4]=0 # tạo hình ảnh đen bên trong   
plt.imshow(toy_image,cmap='gray') # hiển thị hình ảnh đồ chơi
plt.show() 
toy_image
```

We can rescale along a specific axis:

- `fx`: scale factor along the horizontal axis  
- `fy`: scale factor along the vertical axis


The parameter interpolation estimates pixel values based on neighboring pixels. <code>INTER_NEAREST</code> uses the nearest pixel and <code>INTER_CUBIC</code> uses several pixels near the pixel value we would like to estimate.


```python
new_toy = cv2.resize(toy_image,None,fx=2, fy=1, interpolation = cv2.INTER_NEAREST ) # phóng to hình ảnh đồ chơi theo trục x gấp đôi
plt.imshow(new_toy,cmap='gray') # hiển thị hình ảnh đồ chơi mới
plt.show()
```

Consider the following image:


```python
image = cv2.imread("../img/lenna.png") # đọc hình ảnh lenna
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna
plt.show()
```

We can scale the horizontal axis by two and leave the vertical axis as is:


```python
new_image = cv2.resize(image, None, fx=2, fy=1, interpolation=cv2.INTER_CUBIC) # phóng to hình ảnh lenna theo trục x gấp đôi
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape) # hiển thị kích thước hình ảnh cũ và mới
```

In the same manner, we can scale the vertical axis by two:


```python
new_image = cv2.resize(image, None, fx=1, fy=2, interpolation=cv2.INTER_CUBIC) # phóng to hình ảnh lenna theo trục y gấp đôi
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape) # hiển thị kích thước hình ảnh cũ và mới
```

We can scale the horizontal axis and vertical axis by two.


```python
new_image = cv2.resize(image, None, fx=2, fy=2, interpolation=cv2.INTER_CUBIC) # phóng to hình ảnh lenna theo cả 2 trục x và y gấp đôi
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape) # hiển thị kích thước hình ảnh cũ và mới
```

We can also shrink the image by setting the scaling factor to a real number between 0 and 1:


```python
new_image = cv2.resize(image, None, fx=1, fy=0.5, interpolation=cv2.INTER_CUBIC) # thu nhỏ hình ảnh lenna theo trục y bằng nửa
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape) # hiển thị kích thước hình ảnh cũ và mới
```

We can  also specify the number of rows and columns:


```python
rows = 100 # số hàng mới
cols = 200 # số cột mới
```

```python
new_image = cv2.resize(image, (100, 200), interpolation=cv2.INTER_CUBIC) # thay đổi kích thước hình ảnh lenna về 100x200
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
print("old image shape:", image.shape, "new image shape:", new_image.shape) # hiển thị kích thước hình ảnh cũ và mới
```

## Translation


Translation is  when you  shift the location of the image. <code>tx</code> is the number of pixels you shift the location in the horizontal direction and <code>ty</code> is the number of pixels you shift in the vertical direction. You can create the transformation matrix $M$ to shift the image. 

In this example, we shift the image 100 pixels horizontally:


```python
tx = 100 # dịch chuyển theo trục x
ty = 0 # dịch chuyển theo trục y
M = np.float32([[1, 0, tx], [0, 1, ty]]) # ma trận biến đổi dịch chuyển
M # ma trận biến đổi dịch chuyển
```

The shape of the image is given by:


```python
rows, cols, _ = image.shape # lấy kích thước hình ảnh ban đầu
```

We use the function <code>warpAffine</code> from the <code>cv2</code> module. The first input parater is an image array, the second input parameter is the transformation matrix <code>M</code>, and the final input paramter is the length and width of the output image $(cols,rows)$:


```python
new_image = cv2.warpAffine(image, M, (cols, rows)) # áp dụng biến đổi dịch chuyển lên hình ảnh lenna
```

We can plot the image; the portions of the image that do not have any intensities are set to zero:


```python
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới sau khi dịch chuyển
plt.show()
```

We can see some of the original image has been cut off. We can fix this by changing the output image size: <code>(cols + tx,rows + ty)</code>:


```python
new_image = cv2.warpAffine(image, M, (cols + tx, rows + ty)) # áp dụng biến đổi dịch chuyển lên hình ảnh lenna với kích thước mới
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới sau khi dịch chuyển
plt.show()
```

We can shift the image horizontally:


```python
tx = 0 # dịch chuyển theo trục x
ty = 50 # dịch chuyển theo trục y
M = np.float32([[1, 0, tx], [0, 1, ty]]) # ma trận biến đổi dịch chuyển
new_iamge = cv2.warpAffine(image, M, (cols + tx, rows + ty)) # áp dụng biến đổi dịch chuyển lên hình ảnh lenna với kích thước mới
plt.imshow(cv2.cvtColor(new_iamge, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới sau khi dịch chuyển
plt.show()
```

## Rotation 


We can rotate an image by angle θ which is achieved by the Rotation Matrix <code>getRotationMatrix2D</code>.


<p><code>center</code>: Center of the rotation in the source image. We will only use the center of the image.</p>
<p><code>angle</code>: Rotation angle in degrees. Positive values mean counter-clockwise rotation (the coordinate origin is assumed to be the top-left corner).</p>
<p><code>scale</code>: Isotropic scale factor, in this course the value will be one.</p>


We can rotate our toy image by 45 degrees:


```python
theta = 45.0 # góc quay
M = cv2.getRotationMatrix2D(center=(3, 3), angle=theta, scale=1) # ma trận biến đổi quay    
new_toy_image = cv2.warpAffine(toy_image, M, (6, 6)) # áp dụng biến đổi quay lên hình ảnh đồ chơi
```

```python
plot_image(toy_image, new_toy_image, title_1="Orignal", title_2="rotated image") # hiển thị hình ảnh đồ chơi ban đầu và hình ảnh đồ chơi sau khi quay
```

Looking at intensity values, we see that many values have been interpolated:


```python
new_toy_image # hiển thị hình ảnh đồ chơi sau khi quay
```

We can perform the same operation on color images:


```python
cols, rows, _ = image.shape # lấy kích thước hình ảnh ban đầu
```

```python
M = cv2.getRotationMatrix2D(center=(cols // 2 - 1, rows // 2 - 1), angle=theta, scale=1) # ma trận biến đổi quay
new_image = cv2.warpAffine(image, M, (cols, rows)) # áp dụng biến đổi quay lên hình ảnh lenna
```

```python
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới sau khi dịch chuyển
plt.show()
```

# Mathematical Operations 


## Array Operations 


We can perform array operations on an image; Using Python broadcasting, we can add a constant to each pixel's intensity value. 


```python
new_image = image + 20 # tăng độ sáng hình ảnh lenna lên 20 đơn vị

plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show() # hiển thị hình ảnh lenna mới
```

We can also multiply every pixel's intensity value by a constant value.


```python
new_image = 10 * image # tăng độ sáng hình ảnh lenna lên 10 lần
plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
```

We can add the elements of two arrays of equal shape. In this example, we generate an array of random noises with the same shape and data type as our image.


```python
Noise = np.random.normal(0, 20, (rows, cols, 3)).astype(np.uint8) # tạo nhiễu Gaussian
Noise.shape # hiển thị kích thước nhiễu
```

We add the generated noise to the image and plot the result. We see the values that have corrupted the image:


```python
new_image = image + Noise # thêm nhiễu vào hình ảnh lenna

plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
```

At the same time, we can multiply the elements of two arrays of equal shape. We can multiply the random image and the Lenna image and plot the result. 


```python
new_image = image*Noise # nhân nhiễu vào hình ảnh lenna

plt.imshow(cv2.cvtColor(new_image, cv2.COLOR_BGR2RGB)) # hiển thị hình ảnh lenna mới
plt.show()
```

## Matrix Operations 


Grayscale images are matrices. Consider the following grayscale image:


```python
im_gray = cv2.imread('../img/barbara.png', cv2.IMREAD_GRAYSCALE) # đọc hình ảnh barbara dưới dạng ảnh xám
im_gray.shape # hiển thị kích thước hình ảnh xám

plt.imshow(im_gray,cmap='gray') # hiển thị hình ảnh barbara dưới dạng ảnh xám
plt.show()
```

We can apply algorithms designed for matrices.  We can use  Singular Value Decomposition, decomposing our image matrix into  a product of three matrices.


```python
U, s, V = np.linalg.svd(im_gray , full_matrices=True) # phân rã SVD của hình ảnh barbara
```

We see <code>s</code> is not rectangular:


```python
s.shape # hiển thị kích thước vector đặc trưng s
```

We can convert  <code>s</code> to a diagonal matrix <code>S</code>:


```python
S = np.zeros((im_gray.shape[0], im_gray.shape[1])) # tạo ma trận S với kích thước bằng hình ảnh ban đầu
S[:image.shape[0], :image.shape[0]] = np.diag(s) # gán các giá trị trên đường chéo chính của ma trận S bằng vector đặc trưng s
```

We can plot the matrix `U` and `V`:


```python
plot_image(U,V,title_1="Matrix U ",title_2="matrix  V") # hiển thị ma trận U và V
```

We see most of the elements in `S` are zero:


```python
plt.imshow(S,cmap='gray') # hiển thị ma trận S
plt.show()
```

We can find the matrix product of all the matrices. First, we can perform matrix multiplication on `S` and `U` and assign it  to `B` and plot the results: 


```python
B = S.dot(V) # tái tạo hình ảnh bằng cách nhân ma trận S và V
plt.imshow(B,cmap='gray') # hiển thị hình ảnh tái tạo
plt.show()
```

We can find the matrix product of `U`, `S`, and `B`. We see it’s the entire image:


```python
A = U.dot(B) # tái tạo hình ảnh ban đầu bằng cách nhân ma trận U và B
```

```python
plt.imshow(A,cmap='gray') # hiển thị hình ảnh tái tạo
plt.show()
```

It turns out many elements are redundant, so we can eliminate some rows and columns of `S` and `V` and approximate the image by finding the product.


```python
for n_component in [1,10,100,200, 500]: # lặp qua các số thành phần khác nhau
    S_new = S[:, :n_component] # lấy n_component cột đầu tiên của ma trận S
    V_new = V[:n_component, :] # lấy n_component hàng đầu tiên của ma trận V
    A = U.dot(S_new.dot(V_new)) # tái tạo hình ảnh với số thành phần n_component
    plt.imshow(A,cmap='gray') # hiển thị hình ảnh tái tạo
    plt.title("Number of Components:"+str(n_component)) 
    plt.show()
```

We see we only need 100 to 200 Components to represent the image.
