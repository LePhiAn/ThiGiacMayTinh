#  Geometric Transformations


 Geometric transformations allow you to perform different operations like translation, i.e. to shift, reshape and rotate the image.


We can resize an image using the method  <code>resize()</code> of <code>PIL</code> images, which takes the resized image's <code>width</code> and <code>height</code> as paramters. 

Consider the following image:


```python
image = Image.open("../img/lenna.png") # đọc hình ảnh
plt.imshow(image)
plt.show() # hiển thị hình ảnh
```

We can scale the horizontal axis by two and leave the vertical axis as is:


```python
width, height = image.size # lấy kích thước hình ảnh
new_width = 2 * width # thay đổi kích thước
new_hight = height # giữ nguyên chiều cao
new_image = image.resize((new_width, new_hight)) # thay đổi kích thước hình ảnh
plt.imshow(new_image) # hiển thị hình ảnh mới
plt.show()
```

In the same manner, we can scale the vertical axis by two:


```python
new_width = width # giữ nguyên chiều rộng
new_hight = 2 * height # thay đổi chiều cao
new_image = image.resize((new_width, new_hight)) # thay đổi kích thước hình ảnh
plt.imshow(new_image) # hiển thị hình ảnh mới
plt.show()
```

We can double both the width and the height of the image:


```python
new_width = 2 * width # thay đổi chiều rộng
new_hight = 2 * height # thay đổi chiều cao
new_image = image.resize((new_width, new_hight)) # thay đổi kích thước hình ảnh
plt.imshow(new_image) # hiển thị hình ảnh mới
plt.show()
```

We can also shrink the image's width and height both by 1/2:


```python
new_width = width // 2 # thay đổi chiều rộng
new_hight = height // 2 # thay đổi chiều cao

new_image = image.resize((new_width, new_hight)) # thay đổi kích thước hình ảnh
plt.imshow(new_image)
plt.show()
```

## Rotation 


We can rotate an image by angle $\theta$, using the method `rotate`.


We can rotate our toy image by 45 degrees:


```python
theta = 45 # góc quay
new_image = image.rotate(theta) # quay hình ảnh
```

```python
plt.imshow(new_image) # hiển thị hình ảnh sau khi quay  
plt.show()
```

# Mathematical Operations 


## Array Operations 


We can perform array operations on an image; Using Python broadcasting, we can add a constant to each pixel's intensity value. 

Before doing that, we must first we convert the PIL image to a numpy array.


```python
image = np.array(image) # chuyển hình ảnh sang mảng numpy
```

We can then add the constant to the image array:


```python
new_image = image + 20 # tăng độ sáng hình ảnh
plt.imshow(new_image) # hiển thị hình ảnh sau khi tăng độ sáng
plt.show()
```

We can also multiply every pixel's intensity value by a constant value.


```python
new_image = 10 * image # tăng độ tương phản hình ảnh
plt.imshow(new_image) # hiển thị hình ảnh sau khi tăng độ tương phản
plt.show()
```

We can add the elements of two arrays of equal shape. In this example, we generate an array of random noises with the same shape and data type as our image.


```python
Noise = np.random.normal(0, 20, (height, width, 3)).astype(np.uint8) # tạo nhiễu Gaussian
Noise.shape # kiểm tra kích thước nhiễu
```

We add the generated noise to the image and plot the result. We see the values that have corrupted the image:


```python
new_image = image + Noise # thêm nhiễu vào hình ảnh

plt.imshow(new_image) # hiển thị hình ảnh sau khi thêm nhiễu
plt.show()
```

At the same time, we can multiply the elements of two arrays of equal shape. We can multiply the random image and the Lenna image and plot the result. 


```python
new_image = image*Noise # nhân nhiễu vào hình ảnh

plt.imshow(new_image) # hiển thị hình ảnh sau khi nhân nhiễu
plt.show()
```

## Matrix Operations 


Grayscale images are matrices. Consider the following grayscale image:


```python
im_gray = Image.open("../img/barbara.png") # đọc hình ảnh xám
```

Even though the image is gray, it has three channels; we can convert it to a one-channel image.


```python
from PIL import ImageOps # thêm thư viện để xử lý ảnh xám
```

```python
im_gray = ImageOps.grayscale(im_gray) # chuyển hình ảnh sang xám
```

We can convert the PIL image to a numpy array:


```python
im_gray = np.array(im_gray ) # chuyển hình ảnh xám sang mảng numpy
```

```python
plt.imshow(im_gray,cmap='gray') # hiển thị hình ảnh xám
plt.show()
```

We can apply algorithms designed for matrices.  We can use  Singular Value Decomposition, decomposing our image matrix into  a product of three matrices.


```python
U, s, V = np.linalg.svd(im_gray , full_matrices=True) # phân tích SVD
```

We see <code>s</code> is not rectangular:


```python
s.shape # kiểm tra kích thước vector đặc trưng
```

We can convert  <code>s</code> to a diagonal matrix <code>S</code>:


```python
S = np.zeros((im_gray.shape[0], im_gray.shape[1])) # tạo ma trận S với kích thước giống hình ảnh
S[:image.shape[0], :image.shape[0]] = np.diag(s) # điền giá trị vào ma trận S
```

We can plot the matrix U and V:


```python
plot_image(U, V, title_1="Matrix U", title_2="Matrix V") # hiển thị ma trận U và V
```

We see most of the elements in S are zero:


```python
plt.imshow(S, cmap='gray') # hiển thị ma trận S
plt.show()
```

We can find the matrix product of all the  matrices. First, we can perform matrix multiplication on S and U and assign it to `B` and plot the results: 


```python
B = S.dot(V) # tái tạo hình ảnh bằng cách nhân ma trận S và V
plt.imshow(B,cmap='gray') # hiển thị hình ảnh tái tạo
plt.show()
```

We can find the matrix product of `U`, `S`, and `B`. We see it's the entire image:


```python
A = U.dot(B) # tái tạo hình ảnh bằng cách nhân ma trận U và B
```

```python
plt.imshow(A,cmap='gray') # hiển thị hình ảnh tái tạo
plt.show()
```

It turns out that many elements are redundant. We can eliminate some rows and columns of S and V and approximate the image by finding the product:


```python
for n_component in [1,10,100,200, 500]: # thử với các số thành phần khác nhau
    S_new = S[:, :n_component] # tạo ma trận S mới với số thành phần n_component
    V_new = V[:n_component, :] # tạo ma trận V mới với số thành phần n_component
    A = U.dot(S_new.dot(V_new)) # tái tạo hình ảnh với số thành phần n_component
    plt.imshow(A,cmap='gray')
    plt.title("Number of Components:"+str(n_component)) # hiển thị tiêu đề với số thành phần
    plt.show()
```

We see we only need 100 to 200 Components to represent the image.
